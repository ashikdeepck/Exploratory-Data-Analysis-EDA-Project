```python
print("Missing Values:")
print(df.isnull().sum())

print("\nMissing Value Percentage:")
print(((df.isnull().sum() / len(df)) * 100).round(2))
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="Survived")

plt.title("Passenger Survival Count")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="Sex")

plt.title("Gender Distribution of Passengers")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="Age", bins=20, kde=True)

plt.title("Age Distribution of Passengers")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="Sex", hue="Survived")

plt.title("Survival Count by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")
plt.legend(title="Survived", labels=["No", "Yes"])

plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="Pclass", hue="Survived")

plt.title("Survival Count by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.legend(title="Survived", labels=["No", "Yes"])

plt.show()
```

```python
numeric_df = df.select_dtypes(include="number")

correlation_matrix = numeric_df.corr()

print("Correlation Matrix:")
display(correlation_matrix)
```

```python
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Numerical Features")
plt.show()
```

```python
survival_correlation = correlation_matrix["Survived"].sort_values(
    ascending=False
)

print("Correlation with Survival:")
display(survival_correlation)
```

```python
gender_survival = df.groupby("Sex", observed=True)["Survived"].mean() * 100

print("Survival Rate by Gender (%):")
display(gender_survival.round(2))
```

```python
plt.figure(figsize=(7, 5))

sns.barplot(data=df, x="Sex", y="Survived")

plt.title("Survival Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Survival Rate")
plt.show()
```

```python
class_survival = df.groupby("Pclass", observed=True)["Survived"].mean() * 100

print("Survival Rate by Passenger Class (%):")
display(class_survival.round(2))
```

```python

plt.figure(figsize=(7, 5))

sns.barplot(data=df, x="Pclass", y="Survived")

plt.title("Survival Rate by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Average Survival Rate")
plt.show()
```

```python
print("Total Passengers:", len(df))
print("Total Features:", df.shape[1])


print("\nMissing Values:")
print(df.isnull().sum()[df.isnull().sum() > 0])

print("\nSurvival Distribution:")
print(df["Survived"].value_counts().rename(index={0: "Not Survived", 1: "Survived"}))

print("\nSurvival Percentage:")
print((df["Survived"].value_counts(normalize=True) * 100).round(2))

print("\nGender Distribution:")
print(df["Sex"].value_counts())

print("\nPassenger Class Distribution:")
print(df["Pclass"].value_counts().sort_index())

print("\nAverage Age:", round(df["Age"].mean(), 2))
print("Average Fare:", round(df["Fare"].mean(), 2))
```
